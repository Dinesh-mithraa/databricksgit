```python

spark.sql("""select *,current_timestamp as start_date, cast("2100-01-01" as timestamp) as end_date, 'Y' as active_flag  from scd.type2.source""").createOrReplaceTempView("srcdf")
```

```python
%sql
drop table scd.type2.target ;
```

```python
%sql
create table if not exists scd.type2.target
(
  emp_no int,
  emp_name string,
  designation string,
  pass_date date,
  start_date timestamp,
  end_date timestamp,
  active_flag string
)
```

```python
%sql
MERGE INTO scd.type2.target AS trg
USING srcdf AS src
ON src.emp_no = trg.emp_no
   AND trg.active_flag = 'Y'
WHEN MATCHED AND (
     src.emp_name <> trg.emp_name OR 
     src.designation <> trg.designation OR 
     src.pass_date <> trg.pass_date
)
THEN UPDATE SET 
    trg.end_date = current_timestamp(),
    trg.active_flag = 'N'
```

```python
%sql
MERGE INTO scd.type2.target AS trg
USING srcdf AS src
ON src.emp_no = trg.emp_no
   AND trg.active_flag = 'Y'
WHEN MATCHED AND (
     src.emp_name <> trg.emp_name OR 
     src.designation <> trg.designation OR 
     src.pass_date <> trg.pass_date
)
THEN UPDATE SET 
    trg.end_date = current_timestamp(),
    trg.active_flag = 'N'
```
