```python
%sql
create table if not exists scd.type1.source 
(
  emp_no int,
  emp_name string,
  designation string,
  join_date date
)
```

```python
%sql
insert into scd.type1.source values (1,"Dinesh","BCA","2019-01-02"),
                                   (2,"Ramesh","UG","2019-01-02"),
                                   (5,"Kiruthika","ME","2019-01-02"),
                                   (6,"Raaha","BE","2019-01-02")
```

```python
spark.sql("select * from scd.type1.source where join_date='2019-01-02'").createOrReplaceTempView("sdc_source")
```

```python
%sql
select * from sdc_source
```

```python
%sql
create table scd.type1.target as select * from scd.type1.source;
```

```python
%sql
select * from scd.type1.target;
```

```python
%sql

merge into scd.type1.target as tgt
using sdc_source as src
on tgt.emp_no = src.emp_no
when matched and src.join_date > tgt.join_date then update set *
when not matched then insert *
```
